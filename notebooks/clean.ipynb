```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv("../data/WA_Fn-UseC_-Telco-Customer-Churn.csv")
```

```python
df["TotalCharges"] = df["TotalCharges"].replace(" ", np.nan)
df["TotalCharges"] = pd.to_numeric(df["TotalCharges"])
df["TotalCharges"].isnull().sum()
df["TotalCharges"] = df["TotalCharges"].fillna(0)
df.isnull().sum()
```

```python
df.dtypes
```

```python
num_cols = ["tenure", "MonthlyCharges", "TotalCharges"]

for col in num_cols:
    plt.figure(figsize=(6,3))
    sns.boxplot(x=df[col])
    plt.title(f"Boxplot of {col}")
    plt.show()
```

The boxplots do not indicate any abnormal or invalid outliers. The observed values appear to represent genuine customer behavior rather than data entry errors. Therefore no outlier removal was performed and the dataset is considered clean for feature engineering and model development.
