** Business Understanding **

1) Problem Understanding
   
   A telecommunications company is losing customers and wants to understand why. So the goal of this project is to analyze customer data from csv file and identify the factors that are related to churn then build a machine learning model that can predict whether a customer is likely to churn or not.

2) Business Objective

The main objective is to understand customer behavior and discover the reasons behind customer churn to help the company improve customer retention by identifying customers who are at high risk of leaving the service.

3) Target Variable : Churn

The Churn column is the target variable because it represents the output we want to predict.

Possible values:
Yes
No

4) Problem Type

This is a Binary Classification problem because the target variable has only two possible classes: Yes and No.

5) Initial Hypothesis

Before analyzing the dataset, I expect that follwoing things could be the potential reasons of th churn:

Customers with month-to-month contracts may churn more.
Customers with shorter tenure may have a higher chance of churning.
Higher monthly charges may increase the chances of churn.
Customers using fewer telecom services may be more likely to leave.
Contract type and payment method may also influence customer churn.
There may be not enough attraction for new users to pursue and continue and initally they enjoy good servivce at a very good rate.
