** Step 4 – Exploratory Data Analysis (EDA) **

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv("../data/WA_Fn-UseC_-Telco-Customer-Churn.csv")
```

```python
df["Churn"].value_counts()
```

```python
df["Churn"].value_counts(normalize=True)*100
```

```python
sns.countplot(x="Churn", data=df)
plt.title("Customer Churn Distribution")
plt.show()
```

```python
num_cols = ["tenure", "MonthlyCharges"]

for col in num_cols:
    plt.figure(figsize=(6,4))
    sns.histplot(df[col], kde=True)
    plt.title(col)
    plt.show()
```

```python
sns.heatmap(df.corr(numeric_only=True), annot=True, cmap="Blues")
plt.show()
```

```python
sns.countplot(x="Contract", hue="Churn", data=df)
```

```python
sns.countplot(x="InternetService", hue="Churn", data=df)
```

```python
sns.countplot(x="PaymentMethod", hue="Churn", data=df)
plt.xticks(rotation=20)
```

## Customer Churn EDA Summary ##

# Overview

    The exploratory data analysis was performed to understand customer behavior and identify the factors associated with customer churn.The insights from this analysis will be used during data cleaning, feature engineering, and model building.

# Target Variable: Churn Distribution

    Total Customers: 7,043
    Retained Customers: 5,174 (73.5%)
    Churned Customers: 1,869 (26.5%)

    Observation: The dataset has a moderate class imbalance, so evaluation metrics such as F1-Score and ROC-AUC will be more useful than accuracy alone.

# Key Observations

    Customers with less than 12 months tenure are more likely to churn.
    Month-to-month contracts have the highest churn rate.
    Customers using Fiber Optic internet show higher churn than DSL users.
    Customers paying through Electronic Check have higher churn.
    Customers with higher monthly charges tend to churn more frequently.
    Long-term customers (24+ months) and customers with One-Year or Two-Year contracts are less likely to churn.

# Business Insights

| Feature          | Low Churn Risk    | High Churn Risk  | Recommendation                       |
| ---------------- | ----------------- | ---------------- | ------------------------------------ |
| Tenure           | Over 24 months    | Under 12 months  | Improve onboarding for new customers |
| Contract         | One/Two-Year      | Month-to-month   | Encourage long-term contracts        |
| Internet Service | DSL / No Internet | Fiber Optic      | Review pricing and service quality   |
| Payment Method   | Auto-pay          | Electronic Check | Promote automatic payment methods    |


# EDA Conclusion 

    EDA shows that contract type, tenure, monthly charges, internet service, and payment method are the strongest indicators of customer churn. These features will be important during feature engineering and model building
